```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew

# Data
right_skewed_data = [15, 40, 23, 19, 12, 12, 11, 30, 19, 22, 10, 45, 28, 12, 12, 12, 14, 17,
                     16, 13, 19, 12, 13, 15, 16, 25, 12, 17, 19, 90]

left_skewed_data = [35, 38, 90, 50, 85, 83, 87, 62, 64, 90, 90, 80, 70, 72, 74, 75, 76, 78, 80,
                    82, 84, 85, 86, 88, 90, 92, 94, 96, 98]

# Skewness values
print("Right-skewed skewness:", skew(right_skewed_data))
print("Left-skewed skewness:", skew(left_skewed_data))

# ----------------------------------
# Function to plot Histogram + KDE
def plot_hist_kde(data, title, color):
    plt.figure(figsize=(8, 5))

    # Histogram
    plt.hist(data, bins=10, density=True, alpha=0.5,
             color=color, edgecolor='black', label='Histogram')

    # Seaborn KDE
    sns.kdeplot(data, color='green', linewidth=2, label='KDE (Estimated PDF)')

    plt.title(title)
    plt.xlabel("Value")
    plt.ylabel("Probability Density")
    plt.legend()
    plt.show()

# ----------------------------------
# Plot both datasets
plot_hist_kde(right_skewed_data, "Right-Skewed Data: Histogram + KDE", "skyblue")
plot_hist_kde(left_skewed_data, "Left-Skewed Data: Histogram + KDE", "lightcoral")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from statistics import multimode

# Scores dataset
scores = [
    50, 52, 54, 56, 58,
    60, 62, 64, 66, 68,
    70, 72, 74, 75, 76,
    78, 80, 82, 84, 86,
    88, 90, 92, 94, 96,
    54, 56, 58, 60, 62,
    64, 66, 68, 70, 72,
    74, 75, 76, 78, 80,
    82, 84, 86, 88, 90,
    92, 94, 96, 98, 100
]

# Bins
bins = np.arange(50, 105, 5)

# Descriptive statistics
mean_score = np.mean(scores)
median_score = np.median(scores)
mode_score = multimode(scores)

print(f"Mean   : {mean_score}")
print(f"Median : {median_score}")
print(f"Mode   : {mode_score}")

# Histogram only
plt.figure(figsize=(10, 6))
plt.hist(scores, bins=bins, edgecolor='black', alpha=0.7)
plt.title('Histogram of Math Exam Scores')
plt.xlabel('Score Range')
plt.ylabel('Frequency')
plt.xticks(np.arange(50, 105, 5))
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# KDE plot (Estimated Probability Density Function)
plt.figure(figsize=(10, 6))
sns.kdeplot(
    scores,
    color='green',
    linewidth=2,
    fill=True,
    alpha=0.4,
    label='Estimated PDF (KDE)'
)

plt.title('Distribution of Math Exam Scores')
plt.xlabel('Scores')
plt.ylabel('Density')
plt.xticks(np.arange(50, 105, 5))
plt.legend()
plt.show()
```
